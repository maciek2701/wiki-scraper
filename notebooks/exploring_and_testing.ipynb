```python
### Imports
from pathlib import Path

from wikiscraper.scraper.app import WikiScraperApp
from wikiscraper.scraper.client import WikiClient
from wikiscraper.scraper.parser import ArticleParser
```

```python
### 1 krok budowania projektu
repo_root = Path.cwd().parent
rocket_source = r"fixtures\Team_Rocket.html"
path = repo_root / rocket_source

wikiclient = WikiClient()
html = wikiclient.load_html(path)

parser = ArticleParser(html)
summary = parser.extract_summary()

summary = WikiScraperApp().summary_from_local_html(path)
print(summary)
```
